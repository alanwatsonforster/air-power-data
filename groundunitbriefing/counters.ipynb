```python
import glass
import glass.draw

import math
import os

!mkdir -p counters

def drawcounter(
    type, color="white", sighted=False, identified=False, azimuth=None, protectionclass=None,
    name="",
):

    glass.startgamesetup(None, sheets=[["A"]], writelogfiles=False)

    E = glass.setupgroundunit(
        name,
        type=type,
        hexcode="origin",
        color=color,
        counter=True,
        sighted=sighted,
        identified=identified,
        azimuth=azimuth,
    )

    groundunitcountersize = 0.6 * math.sqrt(4 / 3)
    dotspercounter = 300
    dotsperhex = dotspercounter / groundunitcountersize

    glass.draw.startcanvas(
        -0.51 * groundunitcountersize * math.sqrt(4/3),
        -0.51 * groundunitcountersize,
        +0.51 * groundunitcountersize * math.sqrt(4/3),
        +0.51 * groundunitcountersize,
        dotsperhex=dotsperhex,
    )

    #E.takedamage("2D")

    E._draw()

    glass.draw.writecanvastofile("counters/%s.png" % (type))
    glass.draw.writecanvastofile("counters/%s.pdf" % (type), withcreationdate=False)


drawcounter("infantry platoon", identified=True, name="1A2")
drawcounter("light ifv platoon", identified=True, name="M1A2")
```
