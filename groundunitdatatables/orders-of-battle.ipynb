```python
import glass
import glass.draw

import math

!mkdir -p orders-of-battle

xspacefactor = 0.87
yspacefactor = 0.75

def drawstart(rows):
    glass.startgamesetup(None, sheets=[["A"]], writelogfiles=False)

    groundunitcountersize = 0.6 * math.sqrt(4 / 3)
    dotspercounter = 300
    dotsperhex = dotspercounter / groundunitcountersize


    minx = 0
    miny = 0
    maxx = 7
    maxy = rows - 1

    glass.draw.startcanvas(
        minx * xspacefactor - 0.36 * math.sqrt(4/3),
        miny* yspacefactor - 0.36,
        maxx * xspacefactor + 0.36 * math.sqrt(4/3),
        maxy* yspacefactor + 0.36,
        dotsperhex=dotsperhex,
    )

def drawcounter(
    x, y,
    type,
    name="", azimuth=None, 
):


    E = glass.setupgroundunit(
        name,
        type=type,
        hexcode=(x, y),
        color="white",
        counter=True,
        identified=True,
        azimuth=azimuth,
    )
    E._x *= xspacefactor
    E._y *= yspacefactor

    E._draw()
    
def drawend(filename):
    glass.draw.writecanvastofile("orders-of-battle/%s.png" % (filename))


```

```python
battalion = 1
drawstart(5)
for company, y in zip(["A", "B", "C"], [3, 2, 1]):
    drawcounter(0, y, "medium tank platoon", name="1%s%s" %  (company, battalion))
    drawcounter(1, y, "medium tank platoon", name="2%s%s" %  (company, battalion))
    drawcounter(2, y, "medium tank platoon", name="3%s%s" %  (company, battalion))
    drawcounter(3, y, "heavy tank platoon", name="4%s%s" % (company, battalion))
drawcounter(0, 0, "light armored reconnaissance platoon", name="1R%s" % (battalion))
drawcounter(1, 0, "light armored reconnaissance platoon", name="2R%s" % (battalion))
drawcounter(2, 0, "light armored reconnaissance platoon", name="3R%s" % (battalion))
drawcounter(0, 4, "light armored HQ platoon", name="HQ%s" % (battalion))
drawend("british-armoured-regiment-early-1960s")
```
